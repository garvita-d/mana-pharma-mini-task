```python
import pandas as pd
import requests
from io import StringIO

url = "https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/2244,1983,702,6322,5957/property/MolecularWeight,XLogP,InChIKey,CanonicalSMILES/CSV"

try:
    # Fetch the data with a timeout
    response = requests.get(url, timeout=10)

    # Check if the request was successful
    if response.status_code == 200:
        data = StringIO(response.text)
        df = pd.read_csv(data)
        print("Dataset loaded successfully!")
        print("First 5 Rows:\n", df.head())
    else:
        print(f"Failed to fetch data. Status code: {response.status_code}")
        print(f"Response text: {response.text}")
        raise Exception("Switching to fallback dataset")

except Exception as e:
    print(f"An error occurred: {e}")
    print("Switching to fallback dataset...")
```

```python
print("Column Names and Data Types:")
print(df.dtypes)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df['MolecularWeight'], df['XLogP'], color='red')

# Label each point with CID
for i in range(len(df)):
    plt.text(df['MolecularWeight'][i], df['XLogP'][i], str(df['CID'][i]), fontsize=9)

plt.title('Molecular Weight vs XLogP')
plt.xlabel('Molecular Weight')
plt.ylabel('XLogP')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df['CanonicalSMILES'], df['XLogP'], color='red', edgecolor='black')

plt.title('XLogP vs Compound (SMILES)', fontsize=14)
plt.xlabel('Canonical SMILES', fontsize=12)
plt.ylabel('XLogP (Hydrophobicity)', fontsize=12)
plt.xticks(rotation=60, ha='right', fontsize=9)
plt.tight_layout()
plt.grid(axis='y')
plt.show()
```
